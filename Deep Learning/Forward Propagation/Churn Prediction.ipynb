```python
import pandas as pd
```

```python
df = pd.read_csv('Churn_Modelling.csv')
```

```python
df.sample(5)
```

```python
df.drop(columns= ['RowNumber','CustomerId','Surname'] , inplace=True)
```

```python
df['Gender'].value_counts()
```

```python
df = pd.get_dummies(df, columns=['Geography' , 'Gender'] , dtype="int64" , drop_first=True)
```

```python
df
```

```python
X = df.drop(columns=['Exited'])
```

```python
X.sample(5)
```

```python
y = df['Exited'].values
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y , test_size=0.2 , random_state=1)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_trf = scaler.fit_transform(X_train)
X_test_trf = scaler.transform(X_test)
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
X_train_trf.shape
```

```python
model = Sequential()
```

```python
model.add(Dense(5, activation='sigmoid' , input_dim = 11))
model.add(Dense(3, activation='sigmoid' ))
model.add(Dense(1, activation='sigmoid' ))
```

```python
model.summary()
```

```python
model.compile(optimizer='Adam' , loss = 'binary_crossentropy' )
```

```python
model.fit(X_train_trf , y_train , epochs=100, batch_size=50)
```

```python
y_pred = model.predict(X_test_trf)
```

```python
y_pred = (y_pred > 0.5)
```

```python
y_pred = (y_pred > 0.5).astype(int)
```

```python
(y_pred > 0.5).astype(int)
```

```python
y_test
```

```python
y_pred.reshape(y_test.shape)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

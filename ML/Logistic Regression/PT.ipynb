```python
X , y = make_classification(n_samples=100, n_features=4, n_informative=1, n_redundant=0, n_classes=2 , n_clusters_per_class=1, n_repeated=0, random_state=42)
```

```python
class PerceptronTrick:
    def __init__(self, epochs = 100, learning_rate = 0.01):
        self.coeffs = None
        self.intercept = None
        self.epochs = epochs
        self.lr = learning_rate

    def fit(self,X,y):
        self.intercept = 0
        self.coeffs = np.zeros(X.shape[1])
        for i in range(self.epochs):
            # select an random point
            row = np.random.randint(0 , X.shape[0])
            y_cap = step(np.dot(self.coeffs , X[row]) + self.intercept)
            self.coeffs = self.coeffs + self.lr * (y[row] - y_cap) * X[row]
            self.intercept = self.intercept + self.lr * (y[row] - y_cap) 
    def predict(self,X):
        y_prediction = np.dot(X , self.coeffs) + self.intercept
        y_pred = []
        for i in range(X.shape[0]):
            if y_prediction[i] >= 0 :
                y_pred.append(1)
            else:
                y_pred.append(0)
        return y_pred
def step(value):
        return 1 if value >= 0 else 0
    
```

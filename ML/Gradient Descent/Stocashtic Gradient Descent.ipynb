## Our Own SGD Class

```python
# intercept = Wo
class StochasticGradientDescent:
    def __init__(self, epochs = 100, learning_rate = 0.01):
        self.intercept = None
        self.weights = None
        self.lr = learning_rate
        self.epochs = epochs

    def fit (self, X_train, y_train):
        print(X_train.shape[1])
        self.weights = np.ones(X_train.shape[1])
        self.intercept = 0
        print("Initital Weight Shape" , self.weights)
        for i in range(self.epochs):
           for k in range(X_train.shape[0]):
               j = np.random.randint(1 , X_train.shape[0])
               y_cap = self.intercept + np.dot(X_train[j] ,  self.weights)
               self.intercept = self.intercept - 2 * self.lr * (y_train[j] - y_cap) * (-1)
               weights_der = -2  * (y_train[j] - y_cap) * X_train[j]
               self.weights = self.weights - self.lr * weights_der
    def predict(self, X_test):
        return self.intercept + np.dot(X_test , self.weights)
```

```python
sgd = StochasticGradientDescent(epochs=25, learning_rate=0.05)
```

```python
sgd.fit(X_train,y_train)
```

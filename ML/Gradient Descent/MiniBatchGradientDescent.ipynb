```python
from sklearn.datasets import load_diabetes
import numpy as np
from sklearn.linear_model import SGDRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
import pandas as pd
df = pd.DataFrame(X, columns = [f"Feature_{i+1}" for i in range(X.shape[1])])
df["Target"] = y
```

```python
import random
```

```python
# intercept = Wo
class MiniBatchGradientDescent:
    def __init__(self, epochs = 100, learning_rate = 0.01, batch_size = 10):
        self.intercept = None
        self.weights = None
        self.lr = learning_rate
        self.epochs = epochs
        self.batch = batch_size

    def fit (self, X_train, y_train):
        print(X_train.shape[1])
        self.weights = np.ones(X_train.shape[1])
        self.intercept = 0
        print("Initital Weight Shape" , self.weights)
        for i in range(self.epochs):
           for k in range(X_train.shape[0]):
               j = random.sample(range(X_train.shape[0]) , self.batch)
               y_cap = self.intercept + np.dot(X_train[j] ,  self.weights)
               intercept_der =  - 2 * np.mean(y[j] - y_cap)
               self.intercept = self.intercept -  self.lr * intercept_der
               weights_der = -2  * np.dot((y_train[j] - y_cap) , X_train[j]) / self.batch
               self.weights = self.weights - self.lr * weights_der
    def predict(self, X_test):
        return self.intercept + np.dot(X_test , self.weights)
```

```python
gd = MiniBatchGradientDescent()
```

```python
X_train,X_test, y_train,y_test = train_test_split(X,y, test_size=0.2, random_state=13)
```

```python
gd.fit(X_train,y_train)
print( "Intercept: " , gd.intercept)
print( "weights: ",gd.weights)
```

```python
y_pred = gd.predict(X_test)
print(y_pred.shape)
r2_score(y_test, y_pred)
```

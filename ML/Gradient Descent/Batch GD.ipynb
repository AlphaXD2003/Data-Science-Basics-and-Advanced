```python
# intercept = Wo
class BatchGradientDescent:
    def __init__(self, epochs = 100, learning_rate = 0.01):
        self.intercept = None
        self.weights = None
        self.lr = learning_rate
        self.epochs = epochs

    def fit (self, X_train, y_train):
        print(X_train.shape[1])
        self.weights = np.ones(X_train.shape[1])
        self.intercept = 0
        print("Initital Weight Shape" , self.weights)
        for i in range(self.epochs):
            y_cap = self.intercept + np.dot(X_train , self.weights)
            intercept_der = -2 * np.mean(y_train - y_cap)
            self.intercept = self.intercept - self.lr * intercept_der
            weights_der = -2 * np.dot((y_train - y_cap) , X_train) / X_train.shape[0]
            print("weights_der: ", weights_der.shape)
            self.weights = self.weights - self.lr * weights_der
            print(f"{i} Weight Shape: " , self.weights.shape)
            print(f"{i} Weight : " , self.weights)
    def predict(self, X_test):
        return self.intercept + np.dot(X_test , self.weights)
```

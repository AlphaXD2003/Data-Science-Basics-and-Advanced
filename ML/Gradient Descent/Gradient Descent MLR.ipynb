```python
class OurGradientDescent: 
    def __init__(self, learning_rate = 0.01 , epochs = 100):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None
        self.intercept = None

    def fit(self, X, y):
        n_samples, n_features = X.shape
        # Initialization
        self.intercept = 0
        self.weights = np.zeros(n_features)
        print(self.weights.shape)

        # Gradient Descent
        for i in range(self.epochs):
            y_cap = self.intercept + np.dot(X, self.weights)
            i_der = (-2 / n_samples) * np.sum(y - y_cap)
            self.intercept = self.intercept - self.lr * i_der
            w_der = (-2 / n_samples) * np.dot(X.T , (y - y_cap) )
            self.weights = self.weights - self.lr * w_der

    def predict(self,X):
        return np.dot(X, self.weights) + self.intercept
```

```python
gd = OurGradientDescent()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=2)
```

```python
gd.fit(X_train,y_train)
```

```python
y_pred = gd.predict(X_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test, y_pred)
```

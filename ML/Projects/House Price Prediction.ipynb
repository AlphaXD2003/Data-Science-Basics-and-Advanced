```python
# Clean the data
import re

def extract_number(x):
    if pd.isnull(x):
        return None
    nums = re.findall(r'\d+', str(x))
    return float(nums[0]) if nums else None
```

```python
extract_number("9")
```

```python
float("9")
```

```python
df["Bathroom"] = df["Bathroom"].apply(extract_number)
df["Balcony"] = df["Balcony"].apply(extract_number)
df['Carpet Area'] = df['Carpet Area'].apply(extract_number)
df['Super Area'] = df['Super Area'].apply(extract_number)
```

```python
df['Car_Parking_Count'] = df['Car Parking'].str.extract(r'(\d+)').astype(float)
```

```python
df['Car_Parking_Type'] = df['Car Parking'].astype(str).str.extract(r'([A-Za-z]+)')
df['Car_Parking_Type'] = df['Car_Parking_Type'].str.lower()
```

```python
df.drop(columns=["Car Parking" , "Floor"] ,  inplace=True)
```

```python
df.sample(5)
```

```python
def clean_price(x):
    x = str(x).replace(',', '').replace('₹','').strip().lower()

    if 'cr' in x:
        return float(x.replace('cr','')) * 10000000
    elif 'crore' in x:
        return float(x.replace('crore','')) * 10000000
    elif 'lac' in x or 'lakh' in x:
        return float(x.replace('lac','').replace('lakh','')) * 100000
    elif "call for price" in x :
        return 0.0
    else:
        return float(x)
```

```python
clean_price("2.92 Cr")
```

```python
df["Amount(in rupees)"] = df["Amount(in rupees)"].apply(clean_price)
```

```python
# Filling Missing Values
```

```python
num_cols = ['Carpet Area','Super Area','Bathroom','Balcony','Car_Parking_Count' ,'Price (in rupees)']

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.sample(5)
```

```python
cat_cols = ['location', 'Status', 'Furnishing', 'facing', 'Transaction']

df = pd.get_dummies(df, columns=cat_cols, drop_first=True)
```

```python
df = df.astype(int, errors='ignore')
```
